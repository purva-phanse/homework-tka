```python
class Table:
    def __init__(self, size, height, material):
        self.size = size
        self.height = height
        self.material = material

    def toEat(self):
        print("eating on the table")

    def toStudy(self):
        print("Studying on the table")

obj = Table("big",6, "wooden")
obj.toEat()
obj.toStudy()       
```

```python
class Door:
    def __init__(self, material, quality, color):
        self.material = material
        self.quality = quality
        self.color = color

    def toClose(self):
        print("door is closed")

    def toOpen(self):
        print("door is opened")

knob = Door("wood", "high", "brown")
knob.toClose()
knob.toOpen()
```

```python
class Earphone:
    def __init__(self, color, bluetooth, company):
        self.color = color
        self.bluetooth = bluetooth
        self.company = company

    def toVibe(self):
        print("vibing music")

    def toTalk(self):
        print("Talking to someone")

music = Earphone("black", False, "boAt")
music.toVibe()
music.toTalk()
```

```python
class Bag:
    def __init__(self, color, compartments, company):
        self.color = color
        self.compartments= compartments
        self.company = company

    def toCarry(self):
        print("carry laptop")

    def toStore(self):
        print("Store papers")

sack = Bag("black", 3, "Skybags")
sack.toCarry()
sack.toStore()
```
